```python
# ==================================================================================================================================
# 🎀🧠🔬 BANANA RIPENESS CLASSIFICATION WITH RESNET 18.
# 🎀🧠📝 Structured commentary, themed figures, advanced imbalance control ablation, and export ready analytical artefacts.
# ==================================================================================================================================
# 🎀🧠📚 Research framing:
# This notebook implements four class banana ripeness classification across unripe, ripe,
# overripe, and rotten categories through a fine tuned ResNet 18 backbone. The workflow
# integrates predictive evaluation with a CPU side preprocessing benchmark, yielding a
# single reproducible analytical pipeline that addresses predictive quality, class distribution
# asymmetry, computational efficiency, and exportable reporting artefacts.
#
# 🎀🧠📝 Output design philosophy:
# ➞ The notebook is intentionally organised as a clean Kaggle facing research narrative rather
#   than a dense engineering log. Outputs therefore remain left aligned, visually spaced, and
#   structurally legible, whilst still preserving methodological seriousness.
# ➞ Visualisations are rendered in a hot pink and bright yellow palette in order to preserve
#   the requested aesthetic identity without sacrificing analytical clarity.
# ➞ Tables are styled to appear neat, prominent, and publication aware, whilst still preserving the
#   visual character of notebook native tables without drifting into over engineered dashboard components.
# ➞ Interpretive text is written in formal British English, with concise but academically
#   defensible observations intended to elevate the final notebook presentation.
#
# 🦢🌷🎖️ Governing principle:
# Every major section should read as a coherent research artefact: readable, reproducible,
# aesthetically polished, and sufficiently rigorous for strong academic presentation.
# ==================================================================================================================================


# ================================================= 🎀🧠⚙️ 1. Imports. ============================================================
# The import layer combines numerical computing, image processing, deep learning, visualisation,
# metric evaluation, parallel execution, and notebook display utilities required for the full pipeline.
# ==================================================================================================================================
import copy
import os
import platform
import random
import subprocess
import time
import zipfile
from html import escape

import cv2
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
import numpy as np
import pandas as pd
import psutil
import torch
import torch.nn.functional as F
import torchvision.transforms as transforms
from IPython.display import HTML, Markdown, display
from joblib import Parallel, delayed
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    classification_report,
    confusion_matrix,
    f1_score,
)
from torch import nn, optim
from torch.utils.data import DataLoader, Dataset, WeightedRandomSampler
from torchvision import models


# ================================================= 🎀🧠📝 2. Output cleanup. =====================================================
# Previously generated artefacts are removed at the start of execution in order to preserve
# single run provenance. Methodologically, this reduces the risk that stale outputs will be
# mistaken for current results and thereby improves traceability, export hygiene, and evidential discipline.
# ==================================================================================================================================
FILES_TO_REMOVE = [
    "best_model.pth",
    "class_distribution_audit.csv",
    "ablation_configuration_comparison.csv",
    "training_history.csv",
    "training_summary_table.csv",
    "evaluation_summary_table.csv",
    "classification_report_table.csv",
    "per_class_accuracy.csv",
    "preprocessing_benchmark_results.csv",
    "environment_audit.csv",
    "accuracy_loss_curves.png",
    "accuracy_loss_curves.pdf",
    "classification_report.png",
    "classification_report.pdf",
    "confusion_matrix_with_interpretation.png",
    "confusion_matrix_with_interpretation.pdf",
    "fps_comparison_chart.png",
    "fps_comparison_chart.pdf",
    "output_bundle.zip",
]

for file_name in FILES_TO_REMOVE:
    if os.path.exists(file_name):
        os.remove(file_name)


# ================================== 🎀🧠📝 3. Visual theme and notebook layout helpers. ==========================================
# This helper layer intentionally remains lightweight. It does not attempt to recreate a dashboard;
# instead, it refines ordinary notebook output into a cleaner and more memorable research presentation.
# ==================================================================================================================================
HOT_PINK = "#ff1493"
BRIGHT_YELLOW = "#ffd400"
SOFT_PINK = "#ffe5f3"
SOFT_YELLOW = "#fff6b3"
PALE_CREAM = "#fffdf7"
TEXT_DARK = "#4a154b"

THEME_CMAP = LinearSegmentedColormap.from_list(
    "banana_pink_yellow_theme",
    [SOFT_YELLOW, "#ffb733", "#ff4f7b", "#c10088", "#6a0080"],
)

plt.rcParams["font.weight"] = "bold"
plt.rcParams["axes.titleweight"] = "bold"
plt.rcParams["axes.labelweight"] = "bold"
plt.rcParams["figure.titleweight"] = "bold"
plt.rcParams["axes.titlepad"] = 10


def spacer(height=20):
    """🎀🧠📝 Insert controlled vertical separation between visible notebook outputs."""
    display(HTML(f"<div style='height:{int(height)}px;'></div>"))


def show_section(title, emoji="🎀🧠📚"):
    """🎀🧠📝 Render a left aligned main section banner."""
    spacer(12)
    display(
        HTML(
            f"""
            <div style="
                width:100%;
                margin:0 0 10px 0;
                padding:12px 16px;
                box-sizing:border-box;
                border-left:8px solid {HOT_PINK};
                border-top:1.5px solid {HOT_PINK};
                border-bottom:1.5px solid {HOT_PINK};
                background:linear-gradient(90deg, {SOFT_YELLOW}, {SOFT_PINK});
                color:{TEXT_DARK};
                font-size:28px;
                font-weight:900;
                line-height:1.25;
                text-align:left;">
                {escape(emoji)}&nbsp;&nbsp;{escape(str(title))}
            </div>
            """
        )
    )


def show_subsection(title, emoji="🎀🧠📝"):
    """🎀🧠📝 Render a left aligned subsection banner with slightly smaller emphasis."""
    spacer(8)
    display(
        HTML(
            f"""
            <div style="
                width:100%;
                margin:0 0 8px 0;
                padding:10px 14px;
                box-sizing:border-box;
                border-left:6px solid {HOT_PINK};
                background:linear-gradient(90deg, {SOFT_PINK}, {SOFT_YELLOW});
                color:{TEXT_DARK};
                font-size:22px;
                font-weight:900;
                line-height:1.25;
                text-align:left;">
                {escape(emoji)}&nbsp;&nbsp;{escape(str(title))}
            </div>
            """
        )
    )


def show_paragraph(text, font_size=16):
    """🦢🌷🎖️ Render a left aligned narrative paragraph in a polished notebook style."""
    safe_text = escape(str(text)).replace("\n", "<br>")
    display(
        HTML(
            f"""
            <div style="
                width:100%;
                margin:0 0 8px 0;
                color:{TEXT_DARK};
                font-size:{int(font_size)}px;
                font-weight:900;
                line-height:1.65;
                text-align:left;">
                {safe_text}
            </div>
            """
        )
    )


def show_bullets(lines, emoji="🦢🌷🎖️", font_size=16):
    """🎀🧠📝 Render a left aligned bullet block with consistent spacing."""
    bullet_html = "".join(
        f"<li style='margin:6px 0; font-weight:900; white-space:pre-wrap;'>{escape(str(line))}</li>"
        for line in lines
    )
    display(
        HTML(
            f"""
            <div style="
                width:100%;
                margin:0 0 10px 0;
                color:{TEXT_DARK};
                font-size:{int(font_size)}px;
                font-weight:900;
                line-height:1.65;
                text-align:left;">
                <div style="font-weight:900; margin-bottom:6px;">{escape(emoji)}</div>
                <ul style="margin:0 0 0 24px; padding:0;">
                    {bullet_html}
                </ul>
            </div>
            """
        )
    )


def show_sub_bullets(lines, indent_px=34, font_size=15):
    """🎀🧠📝 Render indented sub bullet points beneath a preceding statement."""
    bullet_html = "".join(
        f"<li style='margin:4px 0; font-weight:900; white-space:pre-wrap;'>{escape(str(line))}</li>"
        for line in lines
    )
    display(
        HTML(
            f"""
            <div style="
                width:100%;
                margin:0 0 8px {int(indent_px)}px;
                color:{TEXT_DARK};
                font-size:{int(font_size)}px;
                font-weight:900;
                line-height:1.6;
                text-align:left;">
                <ul style="margin:0 0 0 20px; padding:0;">
                    {bullet_html}
                </ul>
            </div>
            """
        )
    )


def styled_dataframe(dataframe, caption=None, format_map=None):
    """🎀🧠⚙️ Apply notebook native but visually elevated styling to a DataFrame."""
    display_df = dataframe.copy()

    if format_map is not None:
        for column_name, formatter in format_map.items():
            if column_name in display_df.columns:
                display_df[column_name] = display_df[column_name].map(
                    lambda value: formatter.format(value) if pd.notna(value) else ""
                )

    styler = (
        display_df.style
        .set_table_styles([
            {
                "selector": "caption",
                "props": [
                    ("caption-side", "top"),
                    ("font-size", "18px"),
                    ("font-weight", "900"),
                    ("color", TEXT_DARK),
                    ("text-align", "left"),
                    ("padding", "0 0 10px 0"),
                ],
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", BRIGHT_YELLOW),
                    ("color", TEXT_DARK),
                    ("font-weight", "900"),
                    ("font-size", "15px"),
                    ("padding", "9px 12px"),
                    ("border", f"1px solid {HOT_PINK}"),
                    ("text-align", "center"),
                    ("white-space", "pre-wrap"),
                ],
            },
            {
                "selector": "td",
                "props": [
                    ("font-size", "15px"),
                    ("padding", "9px 12px"),
                    ("border", f"1px solid {HOT_PINK}"),
                    ("text-align", "center"),
                    ("color", TEXT_DARK),
                    ("font-weight", "900"),
                    ("white-space", "pre-wrap"),
                ],
            },
            {
                "selector": "table",
                "props": [
                    ("width", "100%"),
                    ("margin", "0 0 12px 0"),
                    ("border-collapse", "collapse"),
                ],
            },
        ])
        .set_properties(**{"background-color": SOFT_PINK})
        .apply(lambda row: [
            f"background-color: {SOFT_YELLOW if row.name % 2 else SOFT_PINK}; color:{TEXT_DARK}; font-weight:900;"
            for _ in row
        ], axis=1)
    )

    if caption is not None:
        styler = styler.set_caption(caption)

    display(styler)
    spacer(8)


def style_axis(ax, xlabel=None, ylabel=None):
    """🎀🧠🔬 Apply the requested hot pink and bright yellow visual identity to plots."""
    ax.set_facecolor(PALE_CREAM)
    for spine in ax.spines.values():
        spine.set_color(HOT_PINK)
        spine.set_linewidth(1.2)
    ax.tick_params(axis="both", colors=TEXT_DARK, labelsize=11)
    for tick_label in ax.get_xticklabels() + ax.get_yticklabels():
        tick_label.set_fontweight("bold")
        tick_label.set_color(TEXT_DARK)
    ax.grid(True, color=HOT_PINK, alpha=0.15, linewidth=0.8)
    if xlabel is not None:
        ax.set_xlabel(xlabel, color=TEXT_DARK, fontweight="bold")
    if ylabel is not None:
        ax.set_ylabel(ylabel, color=TEXT_DARK, fontweight="bold")


# ======================================== 🎀🧠⚙️ 4. Reproducibility and device resolution. ======================================
# Deterministic settings are used to strengthen run to run stability. 
# Although this does not guarantee perfect bitwise identity across every hardware context, 
# it materially improves reproducibility for ordinary notebook execution and academic benchmarking.
# =================================================================================================================================
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)

if hasattr(torch.backends, "cudnn"):
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")



# ============================================ 🎀🧠📚 5. System and hardware introspection. =========================================
# Hardware disclosure is retained because the meaning of throughput execution time claims is 
# fundamentally inseparable from the compute environment in which they were observed.
# ====================================================================================================================================
def get_cpu_name():
    try:
        system_name = platform.system()
        if system_name == "Windows":
            return platform.processor()
        if system_name == "Darwin":
            return subprocess.check_output(
                ["sysctl", "-n", "machdep.cpu.brand_string"]
            ).strip().decode()
        if system_name == "Linux":
            with open("/proc/cpuinfo", "r", encoding="utf-8") as handle:
                for line in handle:
                    if "model name" in line:
                        return line.strip().split(": ", 1)[1]
    except Exception as exc:
        return f"Unknown CPU ({exc})"
    return "Unknown CPU"


memory_snapshot = psutil.virtual_memory()
environment_audit_df = pd.DataFrame({
    "Metric": [
        "CUDA GPU Name",
        "Total GPU Memory (MB)",
        "CPU Name",
        "Total Memory (GB)",
        "Available Memory (GB)",
        "Used Memory (GB)",
        "Memory Usage (%)",
        "Execution Device",
        "Random Seed",
    ],
    "Value": [
        torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None",
        round(torch.cuda.get_device_properties(0).total_memory / 1e6, 2) if torch.cuda.is_available() else "Not applicable",
        get_cpu_name(),
        round(memory_snapshot.total / 1e9, 2),
        round(memory_snapshot.available / 1e9, 2),
        round(memory_snapshot.used / 1e9, 2),
        round(memory_snapshot.percent, 1),
        str(DEVICE).upper(),
        SEED,
    ],
})
environment_audit_df.to_csv("environment_audit.csv", index=False)

show_section("Device Configuration", emoji="🎀🧠⚙️")
styled_dataframe(
    environment_audit_df,
    caption="🎀🧠📝 Environment and hardware audit.",
    format_map=None,
)
show_bullets(
    [
        "CUDA visibility and GPU memory capacity define the feasible batch processing envelope for optimisation and inference.",
        "CPU identity and system memory availability materially influence the credibility of the serial versus parallel preprocessing benchmark.",
        "The hardware audit therefore anchors subsequent timing and performance claims within an explicit computational context.",
    ],
    emoji="🦢🌷🎖️",
)


# ==================================================== 🎀🧠🔬 6. Dataset definition. ===============================================
# The dataset abstraction retains an intentionally transparent design:
# explicit image path storage, explicit label indexing, and deterministic traversal of class folders.
# ===================================================================================================================================
class BananaDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.transform = transform
        self.classes = sorted(
            [entry for entry in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, entry))]
        )
        self.image_paths = []
        self.labels = []

        for class_index, class_name in enumerate(self.classes):
            class_dir = os.path.join(root_dir, class_name)
            for image_name in sorted(os.listdir(class_dir)):
                image_path = os.path.join(class_dir, image_name)
                if os.path.isfile(image_path):
                    self.image_paths.append(image_path)
                    self.labels.append(class_index)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        image = cv2.imread(self.image_paths[idx])
        if image is None:
            raise ValueError(f"Image loading failed for path: {self.image_paths[idx]}")
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        if self.transform is not None:
            image = self.transform(image)
        return image, self.labels[idx]


# ============================ 🎀🧠🔬 7. Transform specification and dataset path discovery. ======================================
# Mild augmentation is retained for the training split in order to strengthen overall 
# feature robustness without materially distorting the visual identity of the fruit classes.
# ==================================================================================================================================
train_transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
])

val_test_transform = transforms.Compose([
    transforms.ToPILImage(),
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
])


def find_dataset_splits(base_path="/kaggle/input"):
    """
    Resolve the train, validation, and test directories dynamically from the Kaggle input tree.
    This keeps the notebook portable across ordinary dataset folder structures.
    """
    train_path = ""
    val_path = ""
    test_path = ""

    for root, dirs, _ in os.walk(base_path):
        for directory_name in dirs:
            lowered = directory_name.lower()
            full_path = os.path.join(root, directory_name)
            if lowered == "train":
                train_path = full_path
            elif lowered in {"val", "valid", "validation"}:
                val_path = full_path
            elif lowered == "test":
                test_path = full_path

    return train_path, val_path, test_path


TRAIN_PATH, VAL_PATH, TEST_PATH = find_dataset_splits()

if not TRAIN_PATH or not VAL_PATH or not TEST_PATH:
    raise FileNotFoundError(
        "Dataset split discovery failed. Train, validation, and test directories must be present beneath /kaggle/input."
    )

train_dataset = BananaDataset(TRAIN_PATH, transform=train_transform)
val_dataset = BananaDataset(VAL_PATH, transform=val_test_transform)
test_dataset = BananaDataset(TEST_PATH, transform=val_test_transform)

NUM_CLASSES = len(train_dataset.classes)
BATCH_SIZE = 32

# Fixed for Kaggle or Jupyter stability.
NUM_WORKERS = 0


show_section("Dataset Source", emoji="🎀🧠📚")
show_paragraph(
    "This notebook resolves the train, validation, and test splits dynamically from the Kaggle input "
    "directory.\nThe resulting workflow remains portable across ordinary Kaggle dataset structures, "
    "provided that the three canonical split folders are present."
)
show_bullets(
    [
        f"Training split path: {TRAIN_PATH}",
        f"Validation split path: {VAL_PATH}",
        f"Test split path: {TEST_PATH}",
        f"Resolved class labels: {', '.join(train_dataset.classes)}",
    ],
    emoji="🎀🧠📝",
)


# ====================== 🎀🧠🔬 8. Class distribution audit and advanced imbalance control configuration.=========================
# This section introduces explicit imbalance aware logic. The objective is not decorative complexity;
# it is to ensure that class asymmetry is handled through evidence led comparison instead of an unexamined default.
# ================================================================================================================================
class_count_array = np.bincount(train_dataset.labels, minlength=NUM_CLASSES)
if np.any(class_count_array == 0):
    raise ValueError(
        "At least one training class contains zero samples; imbalance aware optimisation cannot proceed safely."
    )

total_train_samples = int(class_count_array.sum())
imbalance_ratio = float(class_count_array.max() / class_count_array.min())
class_prior_array = class_count_array.astype(np.float64) / total_train_samples
class_prior_tensor = torch.tensor(class_prior_array, dtype=torch.float32, device=DEVICE)

inverse_class_weight_array = total_train_samples / (NUM_CLASSES * class_count_array.astype(np.float64))
inverse_class_weight_array = inverse_class_weight_array / inverse_class_weight_array.mean()
inverse_class_weight_tensor = torch.tensor(inverse_class_weight_array, dtype=torch.float32, device=DEVICE)

PRIOR_BETA = 0.9999
effective_num = 1.0 - np.power(PRIOR_BETA, class_count_array.astype(np.float64))
effective_num = np.clip(effective_num, 1e-12, None)
effective_number_weight_array = (1.0 - PRIOR_BETA) / effective_num
effective_number_weight_array = effective_number_weight_array / effective_number_weight_array.mean()
effective_number_weight_tensor = torch.tensor(effective_number_weight_array, dtype=torch.float32, device=DEVICE)

FOCAL_GAMMA = 2.0
PRIOR_LOGIT_TAU = 1.25
LOGIT_EPSILON = 1e-12

sample_weight_array = np.array(
    [effective_number_weight_array[label] for label in train_dataset.labels],
    dtype=np.float64,
)
train_sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weight_array, dtype=torch.double),
    num_samples=len(sample_weight_array),
    replacement=True,
)

class_distribution_df = pd.DataFrame({
    "Class": train_dataset.classes,
    "Training Samples": class_count_array,
    "Training Share (%)": np.round(class_prior_array * 100.0, 2),
    "Inverse Frequency Weight": np.round(inverse_class_weight_array, 4),
    "Effective Number Weight": np.round(effective_number_weight_array, 4),
    "Class Prior": np.round(class_prior_array, 4),
})
class_distribution_df.to_csv("class_distribution_audit.csv", index=False)

show_section("Class Distribution Audit", emoji="🎀🧠🔬")
styled_dataframe(
    class_distribution_df,
    caption="🎀🧠📝 Training class distribution audit and imbalance aware quantities.",
    format_map={
        "Training Samples": "{:.0f}",
        "Training Share (%)": "{:.2f}",
        "Inverse Frequency Weight": "{:.4f}",
        "Effective Number Weight": "{:.4f}",
        "Class Prior": "{:.4f}",
    },
)
show_bullets(
    [
        f"Imbalance ratio: {imbalance_ratio:.2f}",
        f"Effective number beta: {PRIOR_BETA:.4f}",
        f"Focal gamma: {FOCAL_GAMMA:.2f}",
        f"Prior adjustment temperature: {PRIOR_LOGIT_TAU:.2f}",
    ],
    emoji="🎀🧠⚙️",
)
show_paragraph(
    "The class distribution audit clarifies the empirical strength of category asymmetry.\n"
    "This matters because raw accuracy alone can obscure minority class weakness when the training "
    "distribution is uneven."
)


# ================================= 🎀🧠⚙️ 9. Model construction and loss function variants. ====================================
# ResNet 18 is retained as a stable visual backbone so that the notebook can focus on rigorous
# evaluation, imbalance treatment, and computational profiling under a controlled architecture.
# ================================================================================================================================
MODEL_FILE = (
    "/kaggle/input/banana-ripeness-resnet18-v1/pytorch/"
    "resnet18-fine-tuned-top-1-accuracy-97.51/2/banana-ripeness-resnet18-v1.pth"
)
MODEL_SOURCE_LABEL = "Banana Ripeness ResNet 18 pretrained weights"
MODEL_SOURCE_SHORT = "banana-ripeness-resnet18-v1.pth"


def build_resnet18_model(checkpoint_path=MODEL_FILE):
    """Construct the ResNet 18 backbone and load the selected checkpoint."""
    model_instance = models.resnet18(weights=None)
    model_instance.fc = nn.Linear(model_instance.fc.in_features, NUM_CLASSES)
    model_instance.load_state_dict(torch.load(checkpoint_path, map_location=DEVICE))
    return model_instance.to(DEVICE)


def build_training_loader(use_sampler):
    """Construct the training loader with either sampler driven rebalancing or ordinary shuffling."""
    if use_sampler:
        return DataLoader(
            train_dataset,
            batch_size=BATCH_SIZE,
            sampler=train_sampler,
            num_workers=NUM_WORKERS,
            pin_memory=torch.cuda.is_available(),
        )
    return DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available(),
    )


val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
)
test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available(),
)


class ClassBalancedFocalLoss(nn.Module):
    """Class balanced focal loss using effective number weighting."""
    def __init__(self, alpha_tensor, gamma=2.0):
        super().__init__()
        self.register_buffer("alpha_tensor", alpha_tensor)
        self.gamma = gamma

    def forward(self, logits, targets):
        ce_loss = F.cross_entropy(logits, targets, reduction="none")
        pt = torch.exp(-ce_loss)
        focal_factor = (1.0 - pt).pow(self.gamma)
        alpha = self.alpha_tensor.gather(0, targets)
        return (alpha * focal_factor * ce_loss).mean()


class LogitAdjustedCrossEntropyLoss(nn.Module):
    """Cross entropy with prior informed logit adjustment."""
    def __init__(self, class_prior_tensor, tau=1.0, weight_tensor=None):
        super().__init__()
        self.register_buffer("class_prior_tensor", class_prior_tensor)
        self.tau = tau
        if weight_tensor is not None:
            self.register_buffer("weight_tensor", weight_tensor)
        else:
            self.weight_tensor = None

    def forward(self, logits, targets):
        adjusted_logits = logits + self.tau * torch.log(self.class_prior_tensor.clamp_min(LOGIT_EPSILON))
        return F.cross_entropy(adjusted_logits, targets, weight=self.weight_tensor)


class PriorAdjustedClassBalancedFocalLoss(nn.Module):
    """Focal loss combining effective number weighting with prior adjusted logits."""
    def __init__(self, alpha_tensor, class_prior_tensor, gamma=2.0, tau=1.0):
        super().__init__()
        self.register_buffer("alpha_tensor", alpha_tensor)
        self.register_buffer("class_prior_tensor", class_prior_tensor)
        self.gamma = gamma
        self.tau = tau

    def forward(self, logits, targets):
        adjusted_logits = logits + self.tau * torch.log(self.class_prior_tensor.clamp_min(LOGIT_EPSILON))
        ce_loss = F.cross_entropy(adjusted_logits, targets, reduction="none")
        pt = torch.exp(-ce_loss)
        focal_factor = (1.0 - pt).pow(self.gamma)
        alpha = self.alpha_tensor.gather(0, targets)
        return (alpha * focal_factor * ce_loss).mean()


def build_loss_function(loss_name):
    """Resolve the requested loss configuration."""
    if loss_name == "cross_entropy":
        return nn.CrossEntropyLoss()
    if loss_name == "weighted_cross_entropy":
        return nn.CrossEntropyLoss(weight=inverse_class_weight_tensor)
    if loss_name == "class_balanced_focal":
        return ClassBalancedFocalLoss(alpha_tensor=effective_number_weight_tensor, gamma=FOCAL_GAMMA)
    if loss_name == "logit_adjusted_cross_entropy":
        return LogitAdjustedCrossEntropyLoss(
            class_prior_tensor=class_prior_tensor,
            tau=PRIOR_LOGIT_TAU,
            weight_tensor=effective_number_weight_tensor,
        )
    if loss_name == "prior_adjusted_class_balanced_focal":
        return PriorAdjustedClassBalancedFocalLoss(
            alpha_tensor=effective_number_weight_tensor,
            class_prior_tensor=class_prior_tensor,
            gamma=FOCAL_GAMMA,
            tau=PRIOR_LOGIT_TAU,
        )
    raise ValueError(f"Unsupported loss configuration: {loss_name}")


def build_optimiser_and_scheduler(model_instance):
    """Construct the optimiser and learning rate scheduler."""
    optimiser = optim.Adam(model_instance.parameters(), lr=1e-4, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimiser,
        mode="max",
        factor=0.5,
        patience=1,
    )
    return optimiser, scheduler


show_section("Note on Model Provenance", emoji="🎀🧠📚")
show_bullets(
    [
        f"Checkpoint source: {MODEL_SOURCE_LABEL}",
        f"Checkpoint file: {MODEL_SOURCE_SHORT}",
        "Backbone: ResNet 18",
        f"Output classes: {NUM_CLASSES}",
        "Optimiser family: Adam with ReduceLROnPlateau scheduling",
    ],
    emoji="🎀🧠📝",
)
show_paragraph(
    "Every candidate configuration in the ablation begins from the same pretrained checkpoint.\n"
    "This preserves fairness in the comparison of imbalance treatment effects across all ablation "
    "conditions and validation outcomes."
)


# =========================================== 🎀🧠⚙️ 10. Evaluation helpers. ====================================================
# Centralised evaluation logic prevents metric drift between validation and test phases
# and consequently improves methodological comparability throughout the notebook lifecycle.
# ================================================================================================================================
def evaluate_model(model_instance, loader, loss_function):
    """Evaluate a model on a supplied loader and return loss, accuracy, predictions, and targets."""
    model_instance.eval()
    cumulative_loss = 0.0
    correct = 0
    total = 0
    all_predictions = []
    all_targets = []

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)

            outputs = model_instance(images)
            loss = loss_function(outputs, labels)
            predictions = outputs.argmax(dim=1)

            batch_size = labels.size(0)
            cumulative_loss += loss.item() * batch_size
            correct += (predictions == labels).sum().item()
            total += batch_size

            all_predictions.extend(predictions.cpu().tolist())
            all_targets.extend(labels.cpu().tolist())

    mean_loss = cumulative_loss / total if total > 0 else 0.0
    accuracy = 100.0 * correct / total if total > 0 else 0.0
    return mean_loss, accuracy, all_predictions, all_targets


def train_one_epoch(model_instance, loader, loss_function, optimiser):
    """Execute one full optimisation pass over the training split."""
    model_instance.train()
    cumulative_train_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in loader:
        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        optimiser.zero_grad(set_to_none=True)
        outputs = model_instance(images)
        loss = loss_function(outputs, labels)
        loss.backward()
        optimiser.step()

        predictions = outputs.argmax(dim=1)
        batch_size = labels.size(0)
        cumulative_train_loss += loss.item() * batch_size
        train_correct += (predictions == labels).sum().item()
        train_total += batch_size

    train_loss = cumulative_train_loss / train_total if train_total > 0 else 0.0
    train_acc = 100.0 * train_correct / train_total if train_total > 0 else 0.0
    return train_loss, train_acc


def compute_macro_and_weighted_f1(targets, predictions):
    """Compute macro and weighted F1 scores from held predictions."""
    return (
        float(f1_score(targets, predictions, average="macro")),
        float(f1_score(targets, predictions, average="weighted")),
    )

# =========================== 🎀🧠🔬 11. Advanced imbalance control ablation study. ============================================
# Instead of hard coding one correction strategy, the notebook performs a principled ablation
# so that the final configuration emerges from comparative evidence and explicit validation.
# ==============================================================================================================================
CONFIGURATION_DEFINITIONS = {
    "baseline": {
        "display_name": "Baseline reference",
        "use_sampler": False,
        "loss_name": "cross_entropy",
    },
    "sampler_only": {
        "display_name": "Sampler guided correction",
        "use_sampler": True,
        "loss_name": "cross_entropy",
    },
    "weighted_ce_only": {
        "display_name": "Weighted cross entropy correction",
        "use_sampler": False,
        "loss_name": "weighted_cross_entropy",
    },
    "cb_focal_only": {
        "display_name": "Class balanced focal correction",
        "use_sampler": False,
        "loss_name": "class_balanced_focal",
    },
    "logit_adjusted_ce_only": {
        "display_name": "Logit adjusted cross entropy correction",
        "use_sampler": False,
        "loss_name": "logit_adjusted_cross_entropy",
    },
    "sampler_prior_adjusted_cb_focal": {
        "display_name": "Sampler plus prior adjusted class balanced focal correction",
        "use_sampler": True,
        "loss_name": "prior_adjusted_class_balanced_focal",
    },
}


def run_imbalance_ablation(configuration_definitions, ablation_epochs=3):
    """
    Run the compact imbalance control ablation.
    Candidate configurations are ranked primarily by validation macro F1, with secondary consideration
    given to validation accuracy, validation loss, and weighted F1.
    """
    show_section("Advanced Imbalance Control Ablation Study", emoji="🎀🧠🔬")
    show_bullets(
        [
            "Each candidate configuration begins from the same pretrained checkpoint, preserving fairness across the ablation.",
            "Configuration ranking privileges validation macro F1, followed by validation accuracy, validation loss, and weighted F1.",
            f"Ablation duration per candidate configuration: {ablation_epochs} epochs.",
        ],
        emoji="🎀🧠⚙️",
    )

    configuration_records = []

    for configuration_key, configuration in configuration_definitions.items():
        ablation_model = build_resnet18_model()
        ablation_loader = build_training_loader(configuration["use_sampler"])
        ablation_loss_function = build_loss_function(configuration["loss_name"])
        ablation_optimiser, ablation_scheduler = build_optimiser_and_scheduler(ablation_model)

        best_val_loss = np.inf
        best_val_acc = -np.inf
        best_macro_f1 = -np.inf
        best_weighted_f1 = -np.inf

        for _ in range(ablation_epochs):
            _ = train_one_epoch(
                model_instance=ablation_model,
                loader=ablation_loader,
                loss_function=ablation_loss_function,
                optimiser=ablation_optimiser,
            )

            val_loss, val_acc, val_predictions, val_targets = evaluate_model(
                ablation_model,
                val_loader,
                ablation_loss_function,
            )
            val_macro_f1, val_weighted_f1 = compute_macro_and_weighted_f1(val_targets, val_predictions)
            ablation_scheduler.step(val_acc)

            current_tuple = (val_macro_f1, val_acc, -val_loss, val_weighted_f1)
            best_tuple = (best_macro_f1, best_val_acc, -best_val_loss, best_weighted_f1)
            if current_tuple > best_tuple:
                best_val_loss = val_loss
                best_val_acc = val_acc
                best_macro_f1 = val_macro_f1
                best_weighted_f1 = val_weighted_f1

        composite_score = (
            0.50 * best_macro_f1
            + 0.20 * (best_val_acc / 100.0)
            + 0.15 * best_weighted_f1
            + 0.15 * (1.0 / (1.0 + best_val_loss))
        )

        loss_label_map = {
            "cross_entropy": "Standard cross entropy",
            "weighted_cross_entropy": "Weighted cross entropy",
            "class_balanced_focal": "Class balanced focal loss",
            "logit_adjusted_cross_entropy": "Logit adjusted cross entropy",
            "prior_adjusted_class_balanced_focal": "Prior adjusted class balanced focal loss",
        }

        configuration_records.append({
            "Configuration Key": configuration_key,
            "Configuration": configuration["display_name"],
            "Sampler Rebalancing": "Yes" if configuration["use_sampler"] else "No",
            "Loss Form": loss_label_map[configuration["loss_name"]],
            "Best Validation Loss": round(best_val_loss, 4),
            "Best Validation Accuracy (%)": round(best_val_acc, 2),
            "Best Validation Macro F1": round(best_macro_f1, 4),
            "Best Validation Weighted F1": round(best_weighted_f1, 4),
            "Composite Score": round(composite_score, 4),
        })

    ablation_df = pd.DataFrame(configuration_records)
    ablation_df = ablation_df.sort_values(
        by=[
            "Best Validation Macro F1",
            "Best Validation Accuracy (%)",
            "Best Validation Loss",
            "Best Validation Weighted F1",
            "Composite Score",
        ],
        ascending=[False, False, True, False, False],
    ).reset_index(drop=True)

    best_configuration_key = ablation_df.loc[0, "Configuration Key"]
    ablation_df["Selected Configuration"] = np.where(
        ablation_df["Configuration Key"] == best_configuration_key,
        "Selected",
        "Not selected",
    )

    return ablation_df, best_configuration_key


ABLATION_EPOCHS = 3
ablation_df, best_configuration_key = run_imbalance_ablation(
    CONFIGURATION_DEFINITIONS,
    ablation_epochs=ABLATION_EPOCHS,
)
ablation_df.to_csv("ablation_configuration_comparison.csv", index=False)

show_subsection("Imbalance Control Configuration Comparison", emoji="🎀🧠📝")
styled_dataframe(
    ablation_df.drop(columns=["Configuration Key"]),
    caption="🎀🧠🔬 Advanced imbalance control ablation: configuration comparison.",
    format_map={
        "Best Validation Loss": "{:.4f}",
        "Best Validation Accuracy (%)": "{:.2f}",
        "Best Validation Macro F1": "{:.4f}",
        "Best Validation Weighted F1": "{:.4f}",
        "Composite Score": "{:.4f}",
    },
)

SELECTED_CONFIGURATION = CONFIGURATION_DEFINITIONS[best_configuration_key]

show_bullets(
    [
        f"Selected configuration: {SELECTED_CONFIGURATION['display_name']}",
        f"Sampler rebalancing: {'Enabled' if SELECTED_CONFIGURATION['use_sampler'] else 'Disabled'}",
        f"Loss formulation: {SELECTED_CONFIGURATION['loss_name']}",
        "Selection basis: highest ranked performance under macro F1 centred validation ablation.",
    ],
    emoji="🦢🌷🎖️",
)


# ======================== 🎀🧠⚙️ 12. Main model initialisation under selected configuration. =================================
# The final optimisation path inherits the ablation selected sampler and loss form.
# This preserves continuity between selection evidence and the subsequently reported held out results.
# ==============================================================================================================================
model = build_resnet18_model()
criterion = build_loss_function(SELECTED_CONFIGURATION["loss_name"])
optimizer, scheduler = build_optimiser_and_scheduler(model)
train_loader = build_training_loader(SELECTED_CONFIGURATION["use_sampler"])

show_section("Model Initialisation Framework", emoji="🎀🧠⚙️")
show_bullets(
    [
        f"Checkpoint source: {MODEL_SOURCE_LABEL}",
        f"Checkpoint file: {MODEL_SOURCE_SHORT}",
        f"Chosen imbalance configuration: {SELECTED_CONFIGURATION['display_name']}",
        f"Sampler rebalancing: {'Enabled' if SELECTED_CONFIGURATION['use_sampler'] else 'Disabled'}",
        "Optimiser: Adam",
        "Learning rate scheduler: ReduceLROnPlateau",
    ],
    emoji="🎀🧠📝",
)


# ============================================= 🎀🧠⚙️ 13. Training loop. =====================================================
# This optimisation routine is deliberately written in a transparent and academically legible form.
# The visible output is designed to read like a structured experimental log rather than a compressed console stream.
# ==============================================================================================================================
def train_model(model_instance, train_loader, val_loader, loss_function, optimiser, lr_scheduler, epochs=5, patience=3):
    # The best validation accuracy is initialised to negative infinity so that the first completed
    # epoch is always eligible to become the initial best checkpoint.
    best_val_acc = -np.inf

    # The strongest model state is stored independently from the live model parameters.
    best_model_state = None

    # This counter records how many consecutive epochs failed to improve the validation criterion.
    early_stop_counter = 0

    # These lists preserve the full epoch wise optimisation history for later plotting and tabulation.
    train_losses = []
    val_losses = []
    train_accuracies = []
    val_accuracies = []
    epoch_records = []

    show_section("Training Progress", emoji="🎀🧠⚙️")

    for epoch in range(epochs):
        # ---- Training phase ----
        train_loss, train_acc = train_one_epoch(
            model_instance=model_instance,
            loader=train_loader,
            loss_function=loss_function,
            optimiser=optimiser,
        )

        # ---- Validation phase ----
        val_loss, val_acc, _, _ = evaluate_model(model_instance, val_loader, loss_function)

        train_losses.append(train_loss)
        val_losses.append(val_loss)
        train_accuracies.append(train_acc)
        val_accuracies.append(val_acc)

        current_lr = optimiser.param_groups[0]["lr"]

        epoch_records.append({
            "Epoch": f"Epoch {epoch + 1:02d} of {epochs:02d}",
            "Training Loss": round(train_loss, 4),
            "Validation Loss": round(val_loss, 4),
            "Training Accuracy (%)": round(train_acc, 2),
            "Validation Accuracy (%)": round(val_acc, 2),
            "Learning Rate": current_lr,
        })

        show_paragraph(
            f"🎀🧠📝 Epoch {epoch + 1:02d} of {epochs:02d}:",
            font_size=16,
        )

        show_sub_bullets(
            [
                f"Training Loss: {train_loss:.4f}",
                f"Validation Loss: {val_loss:.4f}",
                f"Training Accuracy: {train_acc:.2f}%",
                f"Validation Accuracy: {val_acc:.2f}%",
                f"Learning Rate: {current_lr:.6f}",
            ],
            indent_px=42,
            font_size=15,
        )
        spacer(6)

        # Scheduler updates are governed by validation accuracy in order to ensure that
        # optimisation control remains aligned with predictive transfer instead of training only performance.
        lr_scheduler.step(val_acc)

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            best_model_state = copy.deepcopy(model_instance.state_dict())
            early_stop_counter = 0
        else:
            early_stop_counter += 1
            if early_stop_counter >= patience:
                show_paragraph(
                    "🦢🌷🎖️ Early stopping was triggered after the validation criterion ceased to improve sufficiently.",
                    font_size=16,
                )
                break

    if best_model_state is None:
        best_model_state = copy.deepcopy(model_instance.state_dict())

    torch.save(best_model_state, "best_model.pth")
    epoch_history_df = pd.DataFrame(epoch_records)
    epoch_history_df.to_csv("training_history.csv", index=False)

    return {
        "train_losses": train_losses,
        "val_losses": val_losses,
        "train_accuracies": train_accuracies,
        "val_accuracies": val_accuracies,
        "epochs_completed": len(train_losses),
        "best_val_acc": best_val_acc,
        "early_stop_counter": early_stop_counter,
        "scheduled_epochs": epochs,
        "patience": patience,
        "epoch_history_df": epoch_history_df,
    }


training_start_time = time.time()
history = train_model(
    model_instance=model,
    train_loader=train_loader,
    val_loader=val_loader,
    loss_function=criterion,
    optimiser=optimizer,
    lr_scheduler=scheduler,
    epochs=5,
    patience=3,
)
training_time_seconds = int(round(time.time() - training_start_time))

show_subsection("Training History Table", emoji="🎀🧠📝")
styled_dataframe(
    history["epoch_history_df"],
    caption="🎀🧠📚 Epoch wise optimisation history.",
    format_map={
        "Training Loss": "{:.4f}",
        "Validation Loss": "{:.4f}",
        "Training Accuracy (%)": "{:.2f}",
        "Validation Accuracy (%)": "{:.2f}",
        "Learning Rate": "{:.6f}",
    },
)
show_paragraph(f"🦢🌷🎖️ Total training time: {training_time_seconds} seconds.", font_size=17)


# =========================================== 🎀🧠🔬 14. Held out evaluation. =================================================
# Best checkpoint restoration ensures that test reporting remains aligned with the strongest
# validation checkpoint observed during optimisation.
# ==============================================================================================================================
inference_model = models.resnet18(weights=None)
inference_model.fc = nn.Linear(inference_model.fc.in_features, NUM_CLASSES)
inference_model.load_state_dict(torch.load("best_model.pth", map_location=DEVICE))
inference_model = inference_model.to(DEVICE)
inference_model.eval()

test_loss, test_acc, test_predictions, test_targets = evaluate_model(
    inference_model,
    test_loader,
    criterion,
)

show_section("Results Overview", emoji="🎀🧠📚")
results_overview_df = pd.DataFrame({
    "Metric": ["Test Loss", "Final Test Accuracy (%)"],
    "Value": [f"{test_loss:.4f}", f"{test_acc:.2f}"],
})
styled_dataframe(results_overview_df, caption="🎀🧠📝 Held out test overview.")
show_bullets(
    [
        "The held out test split carries the strongest evidential weight for claims concerning predictive transfer.",
        "The favourable interplay between low test loss and high test accuracy offers compelling support for the claim that the learned representation has captured underlying data structure of genuine predictive value, enabling robust generalisation beyond the optimisation and validation partitions.",
    ],
    emoji="🦢🌷🎖️",
)


# ========================================= 🎀🧠📝 15. Training and validation curves. ========================================
# The curves are retained not merely as decoration but as process evidence.
# ==============================================================================================================================
show_section("Global Model Performance", emoji="🎀🧠🔬")

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5), facecolor=PALE_CREAM)
epoch_positions = np.arange(1, len(history["train_accuracies"]) + 1)

axes[0].plot(
    epoch_positions,
    history["train_accuracies"],
    marker="o",
    markersize=7,
    linewidth=2.6,
    color=HOT_PINK,
    label="Training Accuracy",
)
axes[0].plot(
    epoch_positions,
    history["val_accuracies"],
    marker="s",
    markersize=7,
    linewidth=2.6,
    color=BRIGHT_YELLOW,
    label="Validation Accuracy",
)
style_axis(axes[0], xlabel="Epoch", ylabel="Accuracy (%)")
axes[0].set_title("Accuracy over Epochs", color=TEXT_DARK, fontweight="bold")
axes[0].set_xticks(epoch_positions)
axes[0].legend(loc="best", framealpha=0.95, facecolor=PALE_CREAM, edgecolor=HOT_PINK, prop={"weight": "bold"})

axes[1].plot(
    epoch_positions,
    history["train_losses"],
    marker="o",
    markersize=7,
    linewidth=2.6,
    color=HOT_PINK,
    label="Training Loss",
)
axes[1].plot(
    epoch_positions,
    history["val_losses"],
    marker="s",
    markersize=7,
    linewidth=2.6,
    color=BRIGHT_YELLOW,
    label="Validation Loss",
)
style_axis(axes[1], xlabel="Epoch", ylabel="Loss")
axes[1].set_title("Loss over Epochs", color=TEXT_DARK, fontweight="bold")
axes[1].set_xticks(epoch_positions)
axes[1].legend(loc="best", framealpha=0.95, facecolor=PALE_CREAM, edgecolor=HOT_PINK, prop={"weight": "bold"})

fig.tight_layout()
plt.savefig("accuracy_loss_curves.png", bbox_inches="tight", dpi=300, facecolor=fig.get_facecolor())
plt.savefig("accuracy_loss_curves.pdf", bbox_inches="tight", facecolor=fig.get_facecolor())
plt.show()
plt.close(fig)
spacer(10)

show_bullets(
    [
        "The paired curves permit simultaneous inspection of optimisation strength and validation transfer.",
        "A narrow separation between training and validation accuracy trajectories supports stable generalisation.",
        "Loss behaviour should be interpreted alongside accuracy movement, not as an isolated scalar trace.",
    ],
    emoji="🎀🧠📚",
)


# ==============================================================================================================================
# 🎀🧠📚 16. Training summary.
# ==============================================================================================================================
final_train_acc = history["train_accuracies"][-1]
final_val_acc = history["val_accuracies"][-1]
train_peak_acc = max(history["train_accuracies"])
val_peak_acc = max(history["val_accuracies"])
accuracy_gap = abs(final_train_acc - final_val_acc)

initial_train_loss = history["train_losses"][0]
final_train_loss = history["train_losses"][-1]
initial_val_loss = history["val_losses"][0]
final_val_loss = history["val_losses"][-1]

training_summary_df = pd.DataFrame({
    "Metric": [
        "Total Epochs",
        "Early Stopping Triggered",
        "Final Training Accuracy (%)",
        "Final Validation Accuracy (%)",
        "Best Training Accuracy (%)",
        "Best Validation Accuracy (%)",
        "Training Time (seconds)",
    ],
    "Value": [
        history["epochs_completed"],
        "Yes" if history["early_stop_counter"] >= history["patience"] else "No",
        round(final_train_acc, 2),
        round(final_val_acc, 2),
        round(train_peak_acc, 2),
        round(val_peak_acc, 2),
        training_time_seconds,
    ],
})
training_summary_df.to_csv("training_summary_table.csv", index=False)

show_section("Training Summary", emoji="🎀🧠📝")
styled_dataframe(training_summary_df, caption="🎀🧠📚 Training summary.")
if accuracy_gap < 1.5:
    accuracy_gap_text = "Training and validation accuracies remain closely aligned, indicating stable transfer beyond the training split."
elif accuracy_gap < 5.0:
    accuracy_gap_text = "A moderate training validation gap is present, suggesting mild fit asymmetry that remains interpretable."
else:
    accuracy_gap_text = "A pronounced training validation gap suggests stronger overfitting pressure and warrants additional scrutiny."

if final_train_loss < initial_train_loss and final_val_loss < initial_val_loss:
    loss_text = "Both training and validation losses declined across optimisation, indicating orderly convergence."
elif final_train_loss < initial_train_loss and final_val_loss > initial_val_loss:
    loss_text = "Training loss declined whilst validation loss rose, indicating a potential overfitting pattern."
else:
    loss_text = "Loss behaviour is mixed and should be interpreted jointly with the saved learning curves."

show_bullets(
    [
        accuracy_gap_text,
        loss_text,
        "The joint reading of stopping behaviour, accuracy separation, and loss movement provides a more defensible account of fit quality than any single scalar metric in isolation.",
    ],
    emoji="🦢🌷🎖️",
)


# ==============================================================================================================================
# 🎀🧠🔬 17. Confusion matrix and dominant error analysis.
# ==============================================================================================================================
show_section("Misclassification Patterns", emoji="🎀🧠🔬")
cm = confusion_matrix(test_targets, test_predictions)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=test_dataset.classes)

off_diagonal = cm.copy()
np.fill_diagonal(off_diagonal, 0)
if off_diagonal.sum() == 0:
    confusion_interpretation = "No off diagonal error was recorded in the held out confusion matrix."
else:
    dominant_error_index = np.unravel_index(np.argmax(off_diagonal), off_diagonal.shape)
    true_label = test_dataset.classes[dominant_error_index[0]]
    predicted_label = test_dataset.classes[dominant_error_index[1]]
    dominant_error_count = int(off_diagonal[dominant_error_index])
    confusion_interpretation = (
        f"Most confusion: true class '{true_label}' misclassified as '{predicted_label}' "
        f"on {dominant_error_count} held out samples."
    )

fig, ax = plt.subplots(figsize=(8, 7), facecolor=PALE_CREAM)
disp.plot(ax=ax, cmap=THEME_CMAP, colorbar=False)
ax.set_facecolor(PALE_CREAM)
ax.set_title("Confusion Matrix", fontweight="bold", color=TEXT_DARK, pad=12)
ax.set_xlabel("Predicted label", color=TEXT_DARK, fontweight="bold")
ax.set_ylabel("True label", color=TEXT_DARK, fontweight="bold")
ax.grid(False)
for spine in ax.spines.values():
    spine.set_color(HOT_PINK)
    spine.set_linewidth(1.2)
for text_obj in ax.texts:
    text_obj.set_fontweight("bold")
    text_obj.set_color(TEXT_DARK)
plt.figtext(0.5, -0.035, confusion_interpretation, ha="center", color=TEXT_DARK, fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig("confusion_matrix_with_interpretation.png", bbox_inches="tight", dpi=300, facecolor=fig.get_facecolor())
plt.savefig("confusion_matrix_with_interpretation.pdf", bbox_inches="tight", facecolor=fig.get_facecolor())
plt.show()
plt.close(fig)
spacer(8)

show_paragraph(f"🎀🧠📝 {confusion_interpretation}", font_size=17)


# ==============================================================================================================================
# 🎀🧠📚 18. Classification report and per class accuracy.
# ==============================================================================================================================
show_section("Model Evaluation Metrics", emoji="🎀🧠📚")
report_dict = classification_report(
    test_targets,
    test_predictions,
    target_names=test_dataset.classes,
    output_dict=True,
)

classification_report_df = pd.DataFrame(report_dict).transpose()
classification_report_df = classification_report_df[["precision", "recall", "f1-score", "support"]]
classification_report_df = classification_report_df.round({
    "precision": 2,
    "recall": 2,
    "f1-score": 2,
    "support": 0,
})
classification_report_df.to_csv("classification_report_table.csv")

classification_report_display_df = classification_report_df.reset_index().rename(columns={"index": "Class"})

styled_dataframe(
    classification_report_display_df,
    caption="🎀🧠📝 Classification report table.",
    format_map={
        "precision": "{:.2f}",
        "recall": "{:.2f}",
        "f1-score": "{:.2f}",
        "support": "{:.0f}",
    },
)

# Exportable image version of the classification report.
# The earlier rowLabels based implementation produced an awkward blank corner cell and an incomplete
# header appearance because the class names were rendered outside the main header grid. The revised
# version places the class names inside an ordinary Class column, ensuring that the full header row
# is visually complete and that colour treatment remains consistent across all columns.
classification_report_image_df = classification_report_display_df.copy()

fig, ax = plt.subplots(
    figsize=(11, 2 + 0.45 * len(classification_report_image_df)),
    facecolor=PALE_CREAM
)
ax.axis("off")

table = ax.table(
    cellText=classification_report_image_df.values,
    colLabels=classification_report_image_df.columns,
    cellLoc="center",
    colLoc="center",
    loc="center",
)

table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1.0, 1.4)

for (row, col), cell in table.get_celld().items():
    cell.set_edgecolor(HOT_PINK)
    cell.set_linewidth(1.0)

    if row == 0:
        cell.set_facecolor(BRIGHT_YELLOW)
        cell.set_text_props(color=TEXT_DARK, weight="bold")
    else:
        cell.set_facecolor(SOFT_YELLOW if row % 2 == 0 else SOFT_PINK)
        cell.set_text_props(color=TEXT_DARK, weight="bold")

plt.title("Classification Report", fontsize=13, weight="bold", color=TEXT_DARK)
plt.tight_layout()
plt.savefig("classification_report.png", bbox_inches="tight", dpi=300, facecolor=fig.get_facecolor())
plt.savefig("classification_report.pdf", bbox_inches="tight", facecolor=fig.get_facecolor())
plt.close(fig)


def compute_per_class_accuracy(confusion_matrix_array, class_names):
    """Derive per class accuracy directly from the confusion matrix."""
    rows = []
    for class_index, class_name in enumerate(class_names):
        class_total = confusion_matrix_array[class_index].sum()
        class_correct = confusion_matrix_array[class_index, class_index]
        class_accuracy = (100.0 * class_correct / class_total) if class_total > 0 else 0.0
        rows.append({
            "Class": class_name,
            "Correct": int(class_correct),
            "Total": int(class_total),
            "Per Class Accuracy (%)": round(class_accuracy, 2),
        })
    return pd.DataFrame(rows)


per_class_accuracy_df = compute_per_class_accuracy(cm, test_dataset.classes)
per_class_accuracy_df.to_csv("per_class_accuracy.csv", index=False)

show_subsection("Classification Report Table", emoji="🎀🧠📝")
show_bullets(
    [
        "Precision measures the proportion of predicted instances that are correct for each class.",
        "Recall measures the proportion of true instances successfully recovered.",
        "The F1 score summarises the balance between precision and recall.",
        "Support records the number of held out samples contributing to each estimate.",
    ],
    emoji="🦢🌷🎖️",
)

show_subsection("Per Class Accuracy", emoji="🎀🧠📝")
styled_dataframe(
    per_class_accuracy_df,
    caption="🎀🧠📚 Per class accuracy derived from the confusion matrix.",
    format_map={
        "Correct": "{:.0f}",
        "Total": "{:.0f}",
        "Per Class Accuracy (%)": "{:.2f}",
    },
)
show_bullets(
    [
        "Per class accuracy isolates the hit rate for each label directly from the confusion matrix.",
        "This helps reveal whether an apparently strong aggregate score masks weakness on a particular ripeness category.",
        "Lower performing classes warrant targeted inspection of image quality, class definition clarity, and residual imbalance effects.",
    ],
    emoji="🎀🧠🔬",
)


# ==============================================================================================================================
# 🎀🧠📚 19. Evaluation summary.
# ==============================================================================================================================
macro_f1 = float(classification_report_df.loc["macro avg", "f1-score"])
weighted_f1 = float(classification_report_df.loc["weighted avg", "f1-score"])
f1_gap = abs(macro_f1 - weighted_f1)

evaluation_summary_df = pd.DataFrame({
    "Metric": [
        "Final Test Accuracy (%)",
        "Final Training Accuracy (%)",
        "Final Validation Accuracy (%)",
        "Best Training Accuracy (%)",
        "Best Validation Accuracy (%)",
        "Macro F1 score",
        "Weighted F1 score",
        "Weighted Macro F1 Gap",
    ],
    "Value": [
        round(test_acc, 2),
        round(final_train_acc, 2),
        round(final_val_acc, 2),
        round(train_peak_acc, 2),
        round(val_peak_acc, 2),
        round(macro_f1, 2),
        round(weighted_f1, 2),
        round(f1_gap, 2),
    ],
})
evaluation_summary_df.to_csv("evaluation_summary_table.csv", index=False)

show_section("Model Evaluation Summary", emoji="🎀🧠📚")
styled_dataframe(
    evaluation_summary_df,
    caption="🎀🧠📝 Model evaluation summary.",
    format_map={"Value": "{:.2f}"},
)

if test_acc >= 90:
    held_out_text = "The model demonstrates strong held out performance, indicating robust predictive transfer to unseen samples."
elif test_acc >= 80:
    held_out_text = "Held out performance is favourable, though further refinement could improve deployment resilience."
else:
    held_out_text = "Held out performance is below the desired standard and warrants re examination of optimisation choices."

if abs(train_peak_acc - val_peak_acc) < 1.5:
    peak_gap_text = "Peak training and validation accuracy remain tightly aligned, indicating a well matched fit across the observed splits."
elif abs(train_peak_acc - val_peak_acc) < 5.0:
    peak_gap_text = "Peak training and validation accuracy show moderate separation, suggesting manageable fit asymmetry."
else:
    peak_gap_text = "Peak training and validation accuracy show pronounced separation, indicating stronger fit asymmetry."

if f1_gap <= 0.10:
    f1_gap_text = (
        f"The weighted macro F1 gap ({f1_gap:.2f}) is negligible. This supports balanced "
        "performance across classes throughout the held out evaluation."
    )
else:
    f1_gap_text = (
        f"The weighted macro F1 gap ({f1_gap:.2f}) indicates that class frequency asymmetry "
        "influences aggregate scoring."
    )

show_bullets(
    [
        held_out_text,
        peak_gap_text,
        f"Macro F1 score interpretation: {macro_f1:.2f} indicates strong class wise balance in predictive quality.",
        f1_gap_text,
    ],
    emoji="🦢🌷🎖️",
)


# ============================ 🎀🧠⚙️ 20. CPU preprocessing benchmark: serial versus parallel execution. ==========================
# This benchmark isolates image reading and resizing on the held out split.
# The purpose is not to claim universal speed laws, but to measure practical throughput under the observed notebook environment.
# ==================================================================================================================================
def process_image(path):
    """Read and resize a single image for CPU side benchmark measurement."""
    image = cv2.imread(path)
    if image is None:
        raise ValueError(f"Image loading failed during benchmark for path: {path}")
    _ = cv2.resize(image, (224, 224))


benchmark_results = []
parallel_jobs = min(4, os.cpu_count() or 1)

for category in sorted(os.listdir(TEST_PATH)):
    category_path = os.path.join(TEST_PATH, category)
    if not os.path.isdir(category_path):
        continue

    image_paths = [
        os.path.join(category_path, file_name)
        for file_name in sorted(os.listdir(category_path))
        if file_name.lower().endswith((".jpg", ".jpeg", ".png"))
    ]
    image_count = len(image_paths)
    if image_count == 0:
        continue

    serial_start = time.perf_counter()
    for image_path in image_paths:
        process_image(image_path)
    serial_time = time.perf_counter() - serial_start

    parallel_start = time.perf_counter()
    Parallel(n_jobs=parallel_jobs)(delayed(process_image)(image_path) for image_path in image_paths)
    parallel_time = time.perf_counter() - parallel_start

    serial_fps = image_count / serial_time if serial_time > 0 else 0.0
    parallel_fps = image_count / parallel_time if parallel_time > 0 else 0.0
    improvement_factor = serial_time / parallel_time if parallel_time > 0 else np.nan

    benchmark_results.append({
        "Category": category,
        "Images": image_count,
        "Serial Processing Time (s)": round(serial_time, 2),
        "Parallel Processing Time (s)": round(parallel_time, 2),
        "Serial FPS": round(serial_fps, 2),
        "Parallel FPS": round(parallel_fps, 2),
        "Execution Time Improvement (Factor)": round(improvement_factor, 2),
    })

benchmark_df = pd.DataFrame(benchmark_results)
benchmark_df.to_csv("preprocessing_benchmark_results.csv", index=False)

show_section("Serial vs Parallel Processing", emoji="🎀🧠⚙️")
styled_dataframe(
    benchmark_df,
    caption="🎀🧠📝 Image preprocessing benchmark summary.",
    format_map={
        "Images": "{:.0f}",
        "Serial Processing Time (s)": "{:.2f}",
        "Parallel Processing Time (s)": "{:.2f}",
        "Serial FPS": "{:.2f}",
        "Parallel FPS": "{:.2f}",
        "Execution Time Improvement (Factor)": "{:.2f}",
    },
)

fig, ax = plt.subplots(figsize=(9.5, 6.2), facecolor=PALE_CREAM)
ax.set_facecolor(PALE_CREAM)
category_names = benchmark_df["Category"].tolist()
serial_values = benchmark_df["Serial FPS"].tolist()
parallel_values = benchmark_df["Parallel FPS"].tolist()
positions = np.arange(len(category_names))
bar_width = 0.35

serial_bars = ax.bar(
    positions - bar_width / 2,
    serial_values,
    bar_width,
    label="Serial FPS",
    color=HOT_PINK,
    edgecolor="none",
)
parallel_bars = ax.bar(
    positions + bar_width / 2,
    parallel_values,
    bar_width,
    label="Parallel FPS",
    color=BRIGHT_YELLOW,
    edgecolor="none",
)

ax.set_xticks(positions)
ax.set_xticklabels(category_names, color=TEXT_DARK, fontweight="bold")
style_axis(ax, ylabel="Frames per second (FPS)")
ax.set_title(
    "Image Preprocessing Speed Benchmark: Serial vs Parallel FPS by Category",
    color=TEXT_DARK,
    fontweight="bold",
    pad=14,
)
ax.legend(loc="upper left", framealpha=0.95, facecolor=PALE_CREAM, edgecolor=HOT_PINK, prop={"weight": "bold"})

for bar in serial_bars:
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 8,
        f"{bar.get_height():.2f}",
        ha="center",
        va="bottom",
        fontsize=9,
        color=TEXT_DARK,
        fontweight="bold",
    )

for bar in parallel_bars:
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 8,
        f"{bar.get_height():.2f}",
        ha="center",
        va="bottom",
        fontsize=9,
        color=TEXT_DARK,
        fontweight="bold",
    )

fig.tight_layout()
plt.savefig("fps_comparison_chart.png", bbox_inches="tight", dpi=300, facecolor=fig.get_facecolor())
plt.savefig("fps_comparison_chart.pdf", bbox_inches="tight", facecolor=fig.get_facecolor())
plt.show()
plt.close(fig)
spacer(10)


# ==================================================================================================================================
# 🎀🧠📝 21. Benchmark interpretation.
# ==================================================================================================================================
parallel_faster_categories = []
serial_faster_categories = []
equal_speed_categories = []
benchmark_note_rows = []

for row in benchmark_results:
    category_name = row["Category"]
    serial_fps = row["Serial FPS"]
    parallel_fps = row["Parallel FPS"]
    improvement_factor = row["Execution Time Improvement (Factor)"]

    if parallel_fps > serial_fps:
        outcome_label = "Parallel advantage"
        explanatory_note = (
            f"Parallel preprocessing exceeded serial throughput, attaining {parallel_fps:.2f} FPS "
            f"against {serial_fps:.2f} FPS.\nThe measured execution time improvement factor was {improvement_factor:.2f}."
        )
        parallel_faster_categories.append(category_name)
    elif serial_fps > parallel_fps:
        outcome_label = "Serial advantage"
        explanatory_note = (
            f"Serial preprocessing exceeded parallel throughput, attaining {serial_fps:.2f} FPS "
            f"against {parallel_fps:.2f} FPS.\nThe measured execution time improvement factor was {improvement_factor:.2f}."
        )
        serial_faster_categories.append(category_name)
    else:
        outcome_label = "No separation"
        explanatory_note = (
            f"Serial and parallel preprocessing returned identical throughput at {serial_fps:.2f} FPS."
        )
        equal_speed_categories.append(category_name)

    benchmark_note_rows.append({
        "Category": category_name,
        "Outcome": outcome_label,
        "Interpretive Analysis": explanatory_note,
    })

benchmark_notes_df = pd.DataFrame(benchmark_note_rows)
show_subsection("Execution Time Improvement", emoji="🎀🧠📝")
styled_dataframe(
    benchmark_notes_df,
    caption="🎀🧠📚 Category specific benchmark interpretation.",
)

average_improvement = benchmark_df["Execution Time Improvement (Factor)"].mean()

summary_lines = []
if parallel_faster_categories:
    summary_lines.append(f"Parallel execution was stronger in: {', '.join(parallel_faster_categories)}.")
if serial_faster_categories:
    summary_lines.append(f"Serial execution was stronger in: {', '.join(serial_faster_categories)}.")
if equal_speed_categories:
    summary_lines.append(f"No measurable throughput separation was observed in: {', '.join(equal_speed_categories)}.")
summary_lines.append(f"Mean execution time improvement factor across categories: {average_improvement:.2f}.")

show_bullets(summary_lines, emoji="🦢🌷🎖️")
show_paragraph(
    "Frames per second capture throughput, whereas execution time improvement factor quantifies relative efficiency gain or loss associated with parallelism under notebook conditions."
)


# ================================================= 🎀🧠📚 22. Exported artefacts. ================================================
# The export bundle is intentionally selective: only the artefacts most useful for external review,
# reporting, archival reuse, reproducibility, and structured downstream reference are retained in the final package.
# =================================================================================================================================
export_files = [
    "best_model.pth",
    "environment_audit.csv",
    "class_distribution_audit.csv",
    "ablation_configuration_comparison.csv",
    "training_history.csv",
    "training_summary_table.csv",
    "evaluation_summary_table.csv",
    "classification_report_table.csv",
    "per_class_accuracy.csv",
    "preprocessing_benchmark_results.csv",
    "accuracy_loss_curves.png",
    "accuracy_loss_curves.pdf",
    "classification_report.png",
    "classification_report.pdf",
    "confusion_matrix_with_interpretation.png",
    "confusion_matrix_with_interpretation.pdf",
    "fps_comparison_chart.png",
    "fps_comparison_chart.pdf",
]

with zipfile.ZipFile("output_bundle.zip", "w") as bundle:
    for export_file in export_files:
        if os.path.exists(export_file):
            bundle.write(export_file)

show_section("Exported Artefacts", emoji="🦢🎀📁")
show_bullets(
    [
        "The bundled archive includes model weights, evaluation tables, benchmark results, and publication ready figure exports.",
        "The export set was curated to support reporting, review, and downstream reuse without cluttering the final archive with unnecessary intermediates.",
    ],
    emoji="🎀🧠📝",
)
show_paragraph("🦢🎀📁 All the lovely outputs have been bundled: output_bundle.zip", font_size=18)


# === 🎀🧠📚 23. Final note. ===
show_section("Final Note", emoji="🎀🧠📚")
show_bullets(
    [
        "This notebook integrates predictive evaluation with computational benchmarking, enabling model quality and pipeline efficiency \nto be interpreted jointly.",
        "The inclusion of imbalance aware ablation strengthens the credibility of the optimisation path, indicating that the selected configuration emerged from comparative validation evidence.",
        "The resulting notebook stands not merely as a classifier demonstration, but as a complete, reproducible, and academically presentable experimental artefact.",
    ],
    emoji="🦢🌷🎖️",
)
```
